```python
import pandas as pd
import numpy as np
import os
```

```python
# Raw data folder
RAW_PATH = "../data/raw"

# File paths
sales_file = os.path.join(RAW_PATH, "sales.csv")
customers_file = os.path.join(RAW_PATH, "customers.json")
products_file = os.path.join(RAW_PATH, "products.csv")

# Load datasets
sales = pd.read_csv(sales_file)
customers = pd.read_json(customers_file)
products = pd.read_csv(products_file)

print("All datasets loaded successfully.")
```

```python
print("Sales:", sales.shape)
print("Customers:", customers.shape)
print("Products:", products.shape)
```

```python
# DATA PROFILING - PART 1
# Basic Structure & Data Types
datasets = {
    "SALES": sales,
    "CUSTOMERS": customers,
    "PRODUCTS": products
}

for name, df in datasets.items():

    print("\n" + "=" * 60)
    print(f"{name} DATASET")
    print("=" * 60)

    print("\nRows:", df.shape[0])
    print("Columns:", df.shape[1])

    print("\nColumn Names:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)
```

```python
# ============================================
# DATA PROFILING - PART 1
# Missing Value Analysis
# ============================================

for name, df in datasets.items():

    print("\n" + "=" * 60)
    print(f"{name} DATASET")
    print("=" * 60)

    missing = df.isnull().sum()
    missing_pct = (missing / len(df) * 100).round(2)

    missing_summary = pd.DataFrame({
        "Missing Count": missing,
        "Missing %": missing_pct
    })

    display(missing_summary)
```

```python
# ============================================
# DATA PROFILING - PART 1
# Duplicate Analysis
# ============================================

for name, df in datasets.items():

    print("\n" + "=" * 60)
    print(f"{name} DATASET")
    print("=" * 60)

    print("Duplicate rows:", df.duplicated().sum())
```

```python
# ============================================
# KEY ID DUPLICATE CHECK
# ============================================

print("Duplicate SaleIDs:", sales["SaleID"].duplicated().sum())
print("Duplicate CustomerIDs:", customers["CustomerID"].duplicated().sum())
print("Duplicate ProductIDs:", products["ProductID"].duplicated().sum())
```

```python
# ============================================
# DATA PROFILING - PART 2
# Values & Statistics
# ============================================

for name, df in datasets.items():

    print("\n" + "=" * 60)
    print(f"{name} DATASET")
    print("=" * 60)

    # Unique values
    print("\n--- UNIQUE VALUES ---")
    print(df.nunique())

    # Numerical statistics
    numeric_columns = df.select_dtypes(include=np.number).columns

    if len(numeric_columns) > 0:
        print("\n--- NUMERICAL SUMMARY ---")
        display(df[numeric_columns].describe())

    # Categorical values
    categorical_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns

    if len(categorical_columns) > 0:
        print("\n--- CATEGORICAL VALUES ---")

        for column in categorical_columns:
            print(f"\n{column}:")
            print(df[column].value_counts(dropna=False))
```

```python
# ============================================
# FINAL RAW DATA VALIDATION
# ============================================

print("=" * 60)
print("SALES DATA VALIDATION")
print("=" * 60)

# Check ID relationships
print("\n--- REFERENTIAL INTEGRITY ---")

missing_product_ids = ~sales["ProductID"].isin(products["ProductID"])
missing_customer_ids = ~sales["CustomerID"].isin(customers["CustomerID"])

print("Sales with missing ProductID reference:",
      missing_product_ids.sum())

print("Sales with missing CustomerID reference:",
      missing_customer_ids.sum())

# Check numeric values
print("\n--- NUMERIC VALUE CHECKS ---")

print("Negative SalesAmount:",
      (sales["SalesAmount"] < 0).sum())

print("Zero SalesAmount:",
      (sales["SalesAmount"] == 0).sum())

print("Negative Quantity:",
      (sales["Quantity"] < 0).sum())

print("Zero Quantity:",
      (sales["Quantity"] == 0).sum())

# Check timestamp conversion
print("\n--- TIMESTAMP CHECK ---")

timestamp_check = pd.to_datetime(
    sales["Timestamp"],
    errors="coerce"
)

print("Invalid timestamps:",
      timestamp_check.isna().sum())

print("\nValidation completed.")
```

```python
# ============================================
# SALES DATA CLEANING - INITIAL COPY
# ============================================

sales_clean = sales.copy()

print("Original Sales rows:", len(sales))
print("Working Sales rows:", len(sales_clean))
```

```python
# ============================================
# SALES CLEANING - MISSING VALUE ANALYSIS
# ============================================

print("Missing SalesAmount:", sales_clean["SalesAmount"].isna().sum())
print("Missing Quantity:", sales_clean["Quantity"].isna().sum())

# Check whether both are missing in the same transaction
both_missing = (
    sales_clean["SalesAmount"].isna() &
    sales_clean["Quantity"].isna()
).sum()

print("Both SalesAmount and Quantity missing:", both_missing)

# Check each case separately
print("Only SalesAmount missing:",
      (sales_clean["SalesAmount"].isna() &
       sales_clean["Quantity"].notna()).sum())

print("Only Quantity missing:",
      (sales_clean["Quantity"].isna() &
       sales_clean["SalesAmount"].notna()).sum())
```

```python
# ============================================
# SALES CLEANING - QUANTITY ANALYSIS
# ============================================

print("Quantity statistics:")
display(sales_clean["Quantity"].describe())

print("\nQuantity value counts:")
display(sales_clean["Quantity"].value_counts().sort_index())
```

```python
# ============================================
# SALES CLEANING - MISSING QUANTITY BY PRODUCT
# ============================================

missing_quantity = sales_clean[sales_clean["Quantity"].isna()]

print("Missing Quantity records:", len(missing_quantity))

print("\nMissing Quantity by ProductID:")
display(
    missing_quantity["ProductID"]
    .value_counts()
    .sort_index()
)
```

```python
# Checking whether every product with missing quantity has enough non-missing quantity observations

product_quantity_stats = (
    sales_clean.groupby("ProductID")["Quantity"]
    .agg(["count", "median"])
    .reset_index()
)

print("Product-level quantity statistics:")
display(product_quantity_stats)
```

```python
# ============================================
# SALES DATA CLEANING
# ============================================

# 1. Remove records with missing SalesAmount
sales_clean = sales_clean.dropna(subset=["SalesAmount"]).copy()

# 2. Fill missing Quantity using ProductID median
product_quantity_median = (
    sales_clean.groupby("ProductID")["Quantity"]
    .transform("median")
)

sales_clean["Quantity"] = sales_clean["Quantity"].fillna(
    product_quantity_median
)

# 3. Convert Timestamp to datetime
sales_clean["Timestamp"] = pd.to_datetime(
    sales_clean["Timestamp"],
    errors="coerce"
)

print("Sales cleaning completed.")
print("Remaining rows:", len(sales_clean))
print("Missing SalesAmount:", sales_clean["SalesAmount"].isna().sum())
print("Missing Quantity:", sales_clean["Quantity"].isna().sum())
print("Invalid Timestamp:", sales_clean["Timestamp"].isna().sum())
```

```python
# ============================================
# SALES CLEANING - FINAL VALIDATION
# ============================================

print("========== CLEANED SALES VALIDATION ==========")

print("\nDuplicate SaleIDs:",
      sales_clean["SaleID"].duplicated().sum())

print("Missing ProductIDs:",
      sales_clean["ProductID"].isna().sum())

print("Missing CustomerIDs:",
      sales_clean["CustomerID"].isna().sum())

print("Negative SalesAmount:",
      (sales_clean["SalesAmount"] < 0).sum())

print("Zero SalesAmount:",
      (sales_clean["SalesAmount"] == 0).sum())

print("Negative Quantity:",
      (sales_clean["Quantity"] < 0).sum())

print("Zero Quantity:",
      (sales_clean["Quantity"] == 0).sum())

print("Invalid Timestamps:",
      sales_clean["Timestamp"].isna().sum())

print("\nFinal Sales rows:", len(sales_clean))
```

```python
# ============================================
# CUSTOMERS DATA CLEANING - INITIAL COPY
# ============================================

customers_clean = customers.copy()

print("Original Customer rows:", len(customers))
print("Working Customer rows:", len(customers_clean))
```

```python
# ============================================
# CUSTOMER CLEANING - DUPLICATE ANALYSIS
# ============================================

duplicate_customers = customers_clean[
    customers_clean["CustomerID"].duplicated(keep=False)
].sort_values("CustomerID")

print("Duplicate CustomerID records:",
      duplicate_customers["CustomerID"].duplicated().sum())

print("\nDuplicate CustomerID values:")
display(
    duplicate_customers["CustomerID"]
    .value_counts()
    .sort_index()
)

print("\nSample duplicate records:")
display(duplicate_customers.head(20))
```

```python
sales_clean["SaleID"]      # keep as string
sales_clean["CustomerID"]  # keep as string
customers_clean["CustomerID"]  # keep as string
```

```python
# ============================================
# CUSTOMER CLEANING - DUPLICATE CUSTOMER IDs
# ============================================

duplicate_ids = customers_clean[
    customers_clean["CustomerID"].duplicated(keep=False)
].sort_values("CustomerID")

print("Total duplicate CustomerID records:",
      duplicate_ids.shape[0])

print("\nNumber of unique duplicated CustomerIDs:",
      duplicate_ids["CustomerID"].nunique())

print("\nSample duplicate customer records:")
display(duplicate_ids.head(20))
```

```python
# ============================================
# CUSTOMER CLEANING - REMOVE DUPLICATES
# ============================================

before = len(customers_clean)

customers_clean = customers_clean.drop_duplicates(
    subset=["CustomerID"],
    keep="first"
).copy()

after = len(customers_clean)

print("Rows before deduplication:", before)
print("Rows after deduplication:", after)
print("Duplicate CustomerIDs remaining:",
      customers_clean["CustomerID"].duplicated().sum())
```

```python
# ============================================
# CUSTOMER CLEANING - MISSING VALUES
# ============================================

print("Missing values after deduplication:")
print(customers_clean.isnull().sum())
```

```python
# ============================================
# CUSTOMER CLEANING - HANDLE MISSING VALUES
# ============================================

customers_clean["LastName"] = customers_clean["LastName"].fillna("Unknown")
customers_clean["Region"] = customers_clean["Region"].fillna("Unknown")

print("Missing values after treatment:")
print(customers_clean.isnull().sum())
```

```python
# ============================================
# CUSTOMER CLEANING - GENDER STANDARDIZATION
# ============================================

print("Gender values before standardization:")
print(customers_clean["Gender"].value_counts())
```

```python
# ============================================
# CUSTOMER CLEANING - STANDARDIZE GENDER
# ============================================

gender_mapping = {
    "M": "Male",
    "male": "Male",
    "Male": "Male",
    "F": "Female",
    "female": "Female",
    "Female": "Female"
}

customers_clean["Gender"] = customers_clean["Gender"].replace(gender_mapping)

print("Gender values after standardization:")
print(customers_clean["Gender"].value_counts())
```

```python
# ============================================
# CUSTOMER CLEANING - REGION INSPECTION
# ============================================

print("Region values before standardization:")
print(customers_clean["Region"].value_counts(dropna=False))
```

```python
# ============================================
# CUSTOMER CLEANING - STANDARDIZE REGION
# ============================================

region_mapping = {
    "Californiya": "California",
    "california": "California",
    "NY": "New York",
    "Nw York": "New York",
    "New Yorkk": "New York",
    "Texaz": "Texas",
    "Ohho": "Ohio"
}

customers_clean["Region"] = customers_clean["Region"].replace(region_mapping)

print("Region values after standardization:")
print(customers_clean["Region"].value_counts())
```

```python
# ============================================
# FINAL CUSTOMER DATA VALIDATION
# ============================================

print("=" * 60)
print("CLEANED CUSTOMER DATA VALIDATION")
print("=" * 60)

print("\n--- STRUCTURE ---")
print("Final rows:", len(customers_clean))
print("Final columns:", len(customers_clean.columns))

print("\n--- DUPLICATES ---")
print("Duplicate CustomerIDs:",
      customers_clean["CustomerID"].duplicated().sum())
print("Duplicate rows:",
      customers_clean.duplicated().sum())

print("\n--- MISSING VALUES ---")
print(customers_clean.isnull().sum())

print("\n--- STANDARDIZED GENDER ---")
print(customers_clean["Gender"].value_counts())

print("\n--- STANDARDIZED REGION ---")
print(customers_clean["Region"].value_counts())

print("\nCustomer data validation completed.")
```

```python
# ============================================
# PRODUCT CLEANING - CREATE WORKING COPY
# ============================================

products_clean = products.copy()

print("Original Product rows:", len(products))
print("Working Product rows:", len(products_clean))
```

```python
# ============================================
# PRODUCT CLEANING - STRUCTURE CHECK
# ============================================

print("=" * 60)
print("PRODUCT DATA STRUCTURE")
print("=" * 60)

print("\nRows:", products_clean.shape[0])
print("Columns:", products_clean.shape[1])

print("\nColumn Names:")
print(products_clean.columns.tolist())

print("\nData Types:")
print(products_clean.dtypes)
```

```python
# ============================================
# PRODUCT CLEANING - MISSING VALUES
# ============================================

print("Missing values in Product dataset:")
print(products_clean.isnull().sum())
```

```python
# ============================================
# PRODUCT CLEANING - DUPLICATE CHECK
# ============================================

print("Duplicate ProductIDs:",
      products_clean["ProductID"].duplicated().sum())

print("Duplicate product rows:",
      products_clean.duplicated().sum())
```

```python
# ============================================
# PRODUCT CLEANING - CATEGORY INSPECTION
# ============================================

print("Category values:")
print(products_clean["Category"].value_counts())
```

```python
# ============================================
# FINAL PRODUCT DATA VALIDATION
# ============================================

print("=" * 60)
print("CLEANED PRODUCT DATA VALIDATION")
print("=" * 60)

print("\n--- STRUCTURE ---")
print("Final rows:", len(products_clean))
print("Final columns:", len(products_clean.columns))

print("\n--- DUPLICATES ---")
print("Duplicate ProductIDs:",
      products_clean["ProductID"].duplicated().sum())
print("Duplicate rows:",
      products_clean.duplicated().sum())

print("\n--- MISSING VALUES ---")
print(products_clean.isnull().sum())

print("\n--- CATEGORIES ---")
print(products_clean["Category"].value_counts())

print("\nProduct data validation completed.")
```

```python
# ============================================
# ETL - CREATE PROCESSED DATA DIRECTORY
# ============================================

PROCESSED_PATH = "../data/processed"

os.makedirs(PROCESSED_PATH, exist_ok=True)

print("Processed data directory ready:")
print(os.path.abspath(PROCESSED_PATH))
```

```python
# ============================================
# ETL - EXPORT CLEANED DATASETS
# ============================================

sales_clean.to_csv(
    os.path.join(PROCESSED_PATH, "sales_clean.csv"),
    index=False
)

customers_clean.to_csv(
    os.path.join(PROCESSED_PATH, "customers_clean.csv"),
    index=False
)

products_clean.to_csv(
    os.path.join(PROCESSED_PATH, "products_clean.csv"),
    index=False
)

print("Cleaned datasets exported successfully.")
print("\nFiles created:")
print("- sales_clean.csv")
print("- customers_clean.csv")
print("- products_clean.csv")
```

```python
# ============================================
# ETL - VERIFY PROCESSED FILES
# ============================================

processed_files = [
    "sales_clean.csv",
    "customers_clean.csv",
    "products_clean.csv"
]

print("Processed files verification:")
print("=" * 50)

for file in processed_files:
    file_path = os.path.join(PROCESSED_PATH, file)
    df = pd.read_csv(file_path)

    print(f"\n{file}")
    print("Rows:", len(df))
    print("Columns:", len(df.columns))
    print("File exists:", os.path.exists(file_path))

print("\nETL export verification completed.")
```
